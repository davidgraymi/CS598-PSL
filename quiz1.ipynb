```python
import numpy as np

def check_vectors_dependence(*vectors, atol=1e-8):
    """
    Determines if a set of vectors is linearly independent or dependent.
    Pass any number of 1D vectors as arguments.
    """
    # Convert inputs to 2D matrix where each column is a vector
    matrix = np.column_stack([np.asarray(v, dtype=float) for v in vectors])
    num_vectors = len(vectors)
    
    # 1. Quick Check: More vectors than dimensions means automatically dependent
    num_rows, num_cols = matrix.shape
    if num_cols > num_rows:
        return "Linearly Dependent (More vectors than dimensions)"

    # 2. Rank Check
    rank = np.linalg.matrix_rank(matrix, tol=atol)

    if rank < num_vectors:
        return f"Linearly Dependent (Rank {rank} < {num_vectors})"
    else:
        return "Linearly Independent"
```

```python
a1 = np.array([1, 1, 3, 0])
a2 = np.array([1, 2, 0, -1])
a3 = np.array([1, 3, -3, 2])
a4 = np.array([-2, -2, -6, 0])
a5 = np.array([2, 3, 3, -1])

print(check_vectors_dependence(a1, a2, a3))
print(check_vectors_dependence(a2, a3, a4))
print(check_vectors_dependence(a1, a2, a3, a4))
print(check_vectors_dependence(a2, a3))
```
