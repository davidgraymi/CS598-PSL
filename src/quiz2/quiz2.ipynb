```python
names = ["total_labor_hours", "cases_shipped", "indirect_cost_percent", "holiday"]
grocery_df = pd.read_csv('grocery.txt', delimiter='\t', header=None, names=names, dtype={
    'total_labor_hours': int,
    'cases_shipped': int,
    'indirect_cost_percent': float,
    'holiday': bool
})
print(grocery_df.dtypes)
print(grocery_df.describe())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X = grocery_df[["cases_shipped", "indirect_cost_percent", "holiday"]]
y = grocery_df['total_labor_hours']

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Intercept (Beta 0): {model.intercept_:.2f}")
print("Coefficients (Beta 1, Beta 2, Beta 3):")
for feature, coef in zip(X.columns, model.coef_):
    print(f"  {feature}: {coef:.2f}")

print(f"\nMean Squared Error (MSE): {mse:.2f}")
print(f"R-squared Score (R²): {r2:.2f}")

rss = np.sum((y - y_pred)**2)

print(f"RSS: {rss:.2f}")
```

```python
X_reduced = grocery_df[["cases_shipped"]]

model_reduced = LinearRegression()
model_reduced.fit(X_reduced, y)

y_reduced_pred = model_reduced.predict(X_reduced)

mse_reduced = mean_squared_error(y, y_reduced_pred)
r2_reduced = r2_score(y, y_reduced_pred)

print(f"Intercept (Beta 0): {model_reduced.intercept_:.2f}")
print("Coefficients (Beta 1):")
for feature, coef in zip(X_reduced.columns, model_reduced.coef_):
    print(f"  {feature}: {coef:.2f}")

print(f"\nMean Squared Error (MSE): {mse:.2f}")
print(f"R-squared Score (R²): {r2:.2f}")

rss_reduced = np.sum((y - y_reduced_pred)**2)

print(f"RSS: {rss_reduced:.2f}")

n = len(X)
print(n)
q = len(model.coef_) - len(model_reduced.coef_)
print(q)
denom = n - len(model.coef_) - 1
print(denom)
F = ((rss_reduced - rss)/q) / (rss/(denom))

print(f"F-statistic: {F:.4f}")
```
