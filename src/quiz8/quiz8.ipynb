```python
import numpy as np
import sympy as sp
```

```python
def pnew(X, beta):
    return np.exp(X.T @ beta) / (1 + np.exp(X.T @ beta))
```

```python
beta = np.array([-6.0, 0.5, 1.0])
X = np.array([1.0, 3.0, 3.5])
Y = pnew(X, beta)
Y
```

```python
# Define X1 as a real symbol
X1 = sp.Symbol('X1', real=True)

# Define beta and X vectors
beta = sp.Matrix([-6.0, 0.5, 1.0])
X = sp.Matrix([1.0, X1, 3.5])  # Intercept=1.0, X1=hours, X2=3.5 evaluation

# Linear predictor z = X.T @ beta
z = (X.T * beta)[0, 0]

# Logistic probability P(Y = 1 | X)
p = sp.exp(z) / (1 + sp.exp(z))

sp.pprint(p)

# Solve for the boundary where probability is 50% (p == 0.5)
min_hours = sp.solve(sp.Eq(p, 0.5), X1)
print(f"\nMinimum hours required: {min_hours[0]}")

# Or solve the inequality p > 0.5 directly:
inequality_sol = sp.solve(p > 0.5, X1)
print(f"Solution: {inequality_sol}")
```

```python
beta = sp.Matrix([-2.2, 0.15, 1.10])
X = sp.Matrix([1.0, X1, 0.0])
z = (X.T * beta)[0, 0]
p = sp.exp(z) / (1 + sp.exp(z))
sp.pprint(p)

min_hours = sp.solve(sp.Ge(p, 0.8), X1)
print(f"\nMinimum hours required: {min_hours}")

# Or solve the inequality p > 0.5 directly:
inequality_sol = sp.solve(p >= 0.8, X1)
print(f"Solution: {inequality_sol}")
```
